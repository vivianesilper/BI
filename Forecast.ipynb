```python
import pandas as pd
from sqlalchemy import create_engine
# ── 1. Lê e processa o CSV ──────────────────────────────────────────
file_path = r"C:\Users\Viviane Silper\Desktop\Pentaho_projects\Dados1\world_population.csv"

df = pd.read_csv(file_path, sep=';')

for ano in (2023, 2024, 2025):
    col_anterior = "2022 Population" if ano == 2023 else f"{ano - 1} Population"
    df[f"{ano} Population"] = (df[col_anterior] * df["Growth Rate"]).round().fillna(0).astype("int64")
cols = ["Rank", "CCA3", "Country/Territory", "Capital", "Continent",
        "2022 Population", "2023 Population", "2024 Population", "2025 Population",
        "Growth Rate"]

df_final = df[cols].sort_values("2025 Population", ascending=False).reset_index(drop=True)

# ── 2. Conecta no SQL Server ─────────────────────────────────────────
servidor = "DELL-G3-3579"
porta    = 1433
banco    = "Forecast"   # ← altere aqui
usuario  = "sa"         # ← altere aqui
senha    = "123456"           # ← altere aqui

engine = create_engine(
    f"mssql+pyodbc://{'sa'}:{123456789}@{'DELL-G3-3579'}:{1433}/{'Forecast'}"
    f"?driver=ODBC+Driver+17+for+SQL+Server"
)

# ── 3. Salva no SQL Server ────────────────────────────────────────────
df_final.to_sql(
    name      = "predicao_populacao",
    con       = engine,
    if_exists = "replace",
    index     = False
)

print("✅ Tabela salva no SQL Server com sucesso!")
print(f"   {len(df_final)} linhas inseridas.")
```
